## **1. Find Eigenvalues using Colab**

```python
sigmax = np.array([[0, 1], [1, 0]])

display(Math(f"\\sigma_x = {sp.latex(sp.Matrix(sigmax))}"))

print("---------------")

eigenvalues, eigenvectors = np.linalg.eig(sigmax)
eigenvalues = np.round(eigenvalues, 4)
eigenvectors = np.round(eigenvectors, 4)

display(Math(f"\\text{{Eigenvalues: }} {sp.latex(sp.Matrix(eigenvalues))}"))
display(Math(f"\\text{{Eigenvectors: }} {sp.latex(sp.Matrix(eigenvectors))}"))
```

## **4. Vector normalization by hand and using python 2**

```python
# --- Problem 4: check Problem 3 numerically ---
sigmax = np.array([[0, 1], [1, 0]])

# |+>, |-> are the eigenvectors of sigma_x; np.linalg.eig returns them unit-normalized
_, eigvecs = np.linalg.eig(sigmax)
plus, minus = eigvecs[:, [0]], eigvecs[:, [1]]

display(Math(f"\\ket{{+}} = {sp.latex(sp.Matrix(np.round(plus, 4)))}, \\quad "
             f"\\ket{{-}} = {sp.latex(sp.Matrix(np.round(minus, 4)))}"))

# normalization check: <psi|psi> = 1  (vectors are real, so transpose = conjugate transpose)
for name, psi in [("+", plus), ("-", minus)]:
    print(f"<{name}|{name}> = {np.dot(np.transpose(psi), psi)[0, 0]:.4f}")

print("---------------")

# spin up / down projectors from Problem 2: |0><0| and |1><1|  (outer product via np.dot)
zero, one = np.array([[1], [0]]), np.array([[0], [1]])
P_up   = np.dot(zero, np.transpose(zero))
P_down = np.dot(one,  np.transpose(one))

# Born rule: P_psi(m) = <psi| P_m |psi>
for name, psi in [("+", plus), ("-", minus)]:
    p_up   = np.dot(np.transpose(psi), np.dot(P_up,   psi))[0, 0]
    p_down = np.dot(np.transpose(psi), np.dot(P_down, psi))[0, 0]
    display(Math(f"\\text{{P}}_{{{name}}}(\\uparrow) = {p_up:.4f}, \\quad "
                 f"\\text{{P}}_{{{name}}}(\\downarrow) = {p_down:.4f}"))
```

## **7. Finding the Determinants of Pauli Matrices using Colab**

```python
# --- Problem 7: determinants of the Pauli matrices ---
sigmax = np.array([[0,  1], [1,  0]])
sigmay = np.array([[0, -1j], [1j, 0]])
sigmaz = np.array([[1,  0], [0, -1]])

for name, M in [("\\sigma_x", sigmax), ("\\sigma_y", sigmay), ("\\sigma_z", sigmaz)]:
    det = np.linalg.det(M)
    display(Math(f"\\det {name} = {det.real:.4f}"))
```

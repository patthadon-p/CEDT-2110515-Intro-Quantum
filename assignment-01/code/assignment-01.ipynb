## **5. Vector Inner Product Using Python**

```python
v1 = np.array([[np.sqrt(3)], [1]])
v2 = np.array([[1], [1]])
inner_product = np.vdot(v1, v2)

print(f"Inner product: {inner_product:.3f}")
print("v1 Shape:", v1.shape)
print("v2 Shape:", v2.shape)
```

---

## **6. Inner Product with Complex Coefficient 1**

```python
v1 = np.array([[1j], [1]])
v2 = np.array([[-1j], [1]])
inner_result = np.vdot(v1, v2)

print(f"Inner product: {inner_result:.3f}")
```

---

## **7. Inner Product with Complex Coefficient 2**

```python
v1 = np.array([[1 + 2j], [1]])
v2 = np.array([[1 - 1j], [1]])
inner_result = np.vdot(v1, v2)

print(f"Inner product: {inner_result:.3f}")
```

## **Import External Libraries**

```python
import numpy as np
import sympy as sp

from IPython.display import display, Math
from sklearn.preprocessing import normalize
```

```python
BRAKET_01_BASIS = "\\vert 0 \\rangle / \\vert 1 \\rangle"
BRAKET_PM_BASIS = "\\vert + \\rangle / \\vert - \\rangle"
```

---

## **2. Normalization Using Python**

```python
v1 = np.array([[2], [1]])
display(Math(f"\\text{{Original Vector: }} {sp.latex(sp.Matrix(v1))}"))

v2 = np.round(v1 / np.linalg.norm(v1), 4)
display(Math(f"\\text{{Normalized Vector: }} {sp.latex(sp.Matrix(v2))}"))
```

---

## **4. Normalization Using Python with Complex Coefficient**

```python
v1 = np.array([[2], [1+1j]])
v2 = np.round(v1 / np.linalg.norm(v1), 4)
display(Math(f"\\text{{Normalized Vector: }} {sp.latex(sp.Matrix(v2))}"))
```

---

## **7. Vector Normalization by Hand and Using Python**

```python
# v1 = (1, 2) in the |0>/|1> basis
v1_01 = np.array([[1], [2]])

# basis change matrix from |0>/|1> to |+>/|-> coefficients
M = (1 / np.sqrt(2)) * np.array([[1, 1], [1, -1]])

v1_pm = np.round(M @ v1_01, 4)
display(Math(f"v_1 \\text{{ in }} {BRAKET_PM_BASIS} \\text{{ basis (unnormalized):}}"))
display(Math(sp.latex(sp.Matrix(v1_pm.ravel()))))

print("---------------")

v1_pm_normalized = np.round(v1_pm / np.linalg.norm(v1_pm), 4)
display(Math(f"v_1 \\text{{ in }} {BRAKET_PM_BASIS} \\text{{ basis (normalized):}}"))
display(Math(sp.latex(sp.Matrix(v1_pm_normalized.ravel()))))
```

## **9. Vector Normalization by Hand and Using Python 2**

```python
v1 = np.array([[1], [2]])
v1_normalized = np.round(v1 / np.linalg.norm(v1), 4)
display(Math(f"v_1 \\text{{ in }} {BRAKET_01_BASIS} \\text{{ basis (normalized):}}"))
display(Math(sp.latex(sp.Matrix(v1_normalized.ravel()))))

print("---------------")

prob_0 = np.abs(v1_normalized[0, 0]) ** 2
display(Math(f"P(\\vert 0 \\rangle) = {prob_0:.4f}"))
```
